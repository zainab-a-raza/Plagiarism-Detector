```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
import joblib

df = pd.read_csv("C:\\Users\\hp\\Downloads\\ds project final\\features.csv")

X = df[["prefix", "sub10", "lcs", "inverted", "heap"]]
y = df["label"]

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X, y)

joblib.dump(model, "model.pkl")

print("Model saved successfully with heap feature!")
```
